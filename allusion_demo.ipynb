# Allusion Injector

Enter text below and search for allusions using the Allusion Injector engine.

```python
from allusion_engine import get_allusions_for_text

print('Allusion Injector loaded successfully.')
```

```python
import ipywidgets as widgets
from IPython.display import display, HTML

text_input = widgets.Textarea(
    placeholder='Enter a passage of text here...',
    description='Text:',
    layout=widgets.Layout(width='100%', height='180px')
)

find_button = widgets.Button(
    description='Find Allusions',
    button_style='primary',
    icon='search'
)

clear_button = widgets.Button(
    description='Clear',
    icon='trash'
)

output = widgets.Output()

def find_allusions(button):
    with output:
        output.clear_output()

        text = text_input.value.strip()

        if not text:
            print('Please enter some text first.')
            return

        print('Searching...')

        try:
            results = get_allusions_for_text(text)

            if not results:
                print('No allusions were found.')
                return

            print(f'Found {len(results)} result(s).')
            print()

            for number, result in enumerate(results, 1):
                print(f'{number}. {result}')
                print()

        except Exception as error:
            print('An error occurred:')
            print(error)

def clear_text(button):
    text_input.value = ''
    with output:
        output.clear_output()

find_button.on_click(find_allusions)
clear_button.on_click(clear_text)

display(text_input)
display(widgets.HBox([find_button, clear_button]))
display(output)
```
